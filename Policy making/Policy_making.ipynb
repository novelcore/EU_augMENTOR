```python
import os
import json
import openai
from utils.neo4j_connection import neo4j_connection
from utils.policy_recommendations import get_policy_recommendations

# Import configuration
with open("Resources/neo4j_settings.json", "r") as file:
    neo4j_settings = json.load(file)
with open("Resources/openai_settings.json", "r") as file:
    openai_settings = json.load(file)
openai.api_key = openai_settings["api_key"]
# Connection with Neo4j
graph = neo4j_connection(neo4j_settings=neo4j_settings, clean_graph=False)
# Output path
path = "Recommendations"


# Policy making recommendation
# ------------------------------------------------------------------------------------------------------
os.makedirs(path, exist_ok=True)
pilot = "Demo" # Pilot institution name (example)
course_id = ... # Course ID from Moodle (example)
print(f"[INFO] Generating policy recommendations for Institution: {pilot} and course ID: {course_id}")

try:
    response, structure_response = get_policy_recommendations(
        graph=graph,
        pilot=pilot,
        course_id=course_id,
        openai_settings=openai_settings,
    )

    # Open the file in write mode and store the string
    # Export to file
    filename = f"{path}/{pilot}_{course_id}.md"
    with open(filename, "w", encoding="utf-8") as file:
        file.write(response)
    print(f"[INFO] Storing policy recommendations in file: {filename}")
    # Export to JSON
    filename = f"{path}/{pilot}_{course_id}.jsonl"
    with open(filename, "w", encoding="utf-8") as f:
        for record in structure_response:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")
    print(f"[INFO] Storing policy recommendations in JSON format: {filename}")
    # Export to PDF
    filename = f"{path}/{pilot}_{course_id}.pdf"
    print(f"[INFO] Storing policy recommendations in PDF file: {filename}")
except Exception as e:
    print("[ERROR] Recommendations were not generated")
    print(str(e))
```
