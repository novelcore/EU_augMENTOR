```python
import json
from utils.recommendation import Recommendations
from utils.utils import color
from rich.console import Console
from rich.markdown import Markdown
# Create a console object
console = Console()

# Import configuration
with open("Resources/neo4j_settings.json", "r") as file:
    neo4j_settings = json.load(file)
with open("Resources/openai_settings.json", "r") as file:
    openai_settings = json.load(file)

Recommender = Recommendations(
    neo4j_settings=neo4j_settings,
    openai_settings=openai_settings,
    cypher_corrector=True,
    recommendations_file="...",
    verbose=True,
)

# Set settings
course_id = 4
user_settings = {"title": "Educator", "username": "..."}
configuration = {
    "course_id": course_id,
    "user_settings": user_settings,
    "language": "greek",
    "include_guidelines": False
}
```

```python
question = """Give me information about creativity and critical thinking of the class for all modules"""
configuration = {
    "course_id": course_id,
    "user_settings": user_settings,
    "language": "greek",
    "include_guidelines": False
}

Recommender._verbose = True
response, configuration = Recommender.query(configuration, question)
console.print(Markdown(markup=response))
```

```python
if configuration['guidelines']:
    console.print(Markdown(markup=configuration['guidelines']))
```

```python
Recommender._verbose = False
d_recommendations = Recommender.recommendation(configuration=configuration)

for receiver, value in d_recommendations.items():
    console.print(Markdown(markup=value))
    print(color.CYAN + "To:" + color.END, receiver)
```
